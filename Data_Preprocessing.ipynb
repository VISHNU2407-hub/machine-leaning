```python
import pandas as pd
#missing values
data={
    "Age":[25,30,None,28,35],
    "Salary":[50000,60000,55000,None,70000]
}
df=pd.DataFrame(data)
print(df)
```

```python
#fill the missing values using mean
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Salary']=df["Salary"].fillna(df['Salary'].mean())
print(df)
```

```python
#dropping the duplicates
df=pd.concat([df,df.iloc[[0]]])
print(df)
df=df.drop_duplicates()
print(df)
```

```python
#label encoding
from sklearn.preprocessing  import LabelEncoder
data={
    "Gender":["Apple","Mango","Apple","Mango"]
}
df=pd.DataFrame(data)
print(df)
encoder=LabelEncoder()
df["Gender"]=encoder.fit_transform(df['Gender'])
print(df)
```

```python
#one-Hot encoding
import pandas as pd
df=pd.DataFrame({"Color":['Red',"Green","blue"]})
print(df)
encoded=pd.get_dummies(df)#get dummies creates a unique coloums for each one
print(encoded)
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
import pandas as pd
df=pd.DataFrame({
    "age":[20,25,30,35],
    "Salary":[20000,50000,80000,120000]
})
scaler=StandardScaler()
scaled_data=scaler.fit_transform(df)
print(scaled_data)
```
